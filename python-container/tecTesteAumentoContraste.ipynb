# Técnicas para Validar o Aumento de Contraste
as técnicas apresentadas são:
* Histograma
* Perfil de Linha
* GLCM

## obs: importações

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Histograma

```python
# Carregar imagem em escala de cinza
img = cv2.imread("P_1.jpg", cv2.IMREAD_GRAYSCALE)

# Plotar histograma
plt.figure()
plt.title("Histograma")
plt.xlabel("Intensidade (0-255)")
plt.ylabel("Frequência")
plt.hist(img.ravel(), bins=256, range=(0, 256))
plt.show()
```

## Perfil de Linha

```python
# Carregar imagem em escala de cinza
img = cv2.imread("P_1.jpg", cv2.IMREAD_GRAYSCALE)

# Definir a linha (meio da imagem)
linha = img[img.shape[0] // 2, :]

# Plotar o perfil de linha
plt.figure()
plt.title("Perfil de Linha (linha central)")
plt.xlabel("Posição (pixels)")
plt.ylabel("Intensidade")
plt.plot(linha)
plt.show()
```
