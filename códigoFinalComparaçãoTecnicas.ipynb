# código gráfico desvio padrão

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# 1. DADOS GLCM (valores em %)
# ============================================================

# HE
data_he = {
    "Imagem": ["Normal 1", "Normal 2", "Glioma", "Meningioma"],
    "Contrast": [-66.36, -66.87, 64.35, -24.28],
    "Homogeneity": [5.77, -2.94, -6.22, 1.18],
    "Energy": [-5.83, -14.36, -2.04, -4.67],
    "Correlation": [4.40, 5.54, 1.27, 1.15]
}

# CLAHE
data_clahe = {
    "Imagem": ["Normal 1", "Normal 2", "Glioma", "Meningioma"],
    "Contrast": [-83.97, -78.19, -30.03, -27.35],
    "Homogeneity": [17.95, 18.48, -3.36, 7.97],
    "Energy": [4.06, 2.75, -1.21, 0.65],
    "Correlation": [4.76, 5.50, 1.20, 0.87]
}

# ============================================================
# ESPAÇO RESERVADO PARA DIP
# ============================================================

data_dip = {
    "Imagem": ["Normal 1", "Normal 2", "Glioma", "Meningioma"],
    "Contrast": [-70.97, -74.94, -38.02, -39.58],        # preencher
    "Homogeneity": [-29.41, -9.23, -41.53, -32.72],     # preencher
    "Energy": [-31.52, -0.72, -39.54, -32.75],          # preencher
    "Correlation": [3.69, 4.66, 0.52, 0.68]      # preencher
}

# ============================================================
# ESPAÇO RESERVADO PARA ZERO-DCE
# ============================================================

data_zero_dce = {
    "Imagem": ["Normal 1", "Normal 2", "Glioma", "Meningioma"],
    "Contrast": [325.47, 548.75, 1093.43, 387.50],        # preencher
    "Homogeneity": [169.80, 164.85, 41.52, 45.34],     # preencher
    "Energy": [411.23, 511.77, 52.76, 76.05],          # preencher
    "Correlation": [-4.95, -11.45, -0.37, -0.49]      # preencher
}

# ============================================================
# 2. DATAFRAMES
# ============================================================

df_he = pd.DataFrame(data_he)
df_clahe = pd.DataFrame(data_clahe)

# Esses só funcionarão quando os dados forem inseridos
df_dip = pd.DataFrame(data_dip)
df_zero_dce = pd.DataFrame(data_zero_dce)

metrics = ["Contrast", "Homogeneity", "Energy", "Correlation"]

# ============================================================
# 3. DESVIO PADRÃO
# ============================================================

std_he = df_he[metrics].std()
std_clahe = df_clahe[metrics].std()

# Inicialmente vazios (evita erro e já deixa pronto)
std_dip = None
std_zero_dce = None

if not df_dip[metrics].isnull().values.any() and len(df_dip[metrics]) > 0:
    std_dip = df_dip[metrics].std()

if not df_zero_dce[metrics].isnull().values.any() and len(df_zero_dce[metrics]) > 0:
    std_zero_dce = df_zero_dce[metrics].std()

# ============================================================
# 4. GRÁFICO COMPARATIVO
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(std_he.index, std_he.values, marker='o', label="HE")
plt.plot(std_clahe.index, std_clahe.values, marker='o', label="CLAHE")

if std_dip is not None:
    plt.plot(std_dip.index, std_dip.values, marker='o', label="DIP")

if std_zero_dce is not None:
    plt.plot(std_zero_dce.index, std_zero_dce.values, marker='o', label="Zero-DCE")

plt.title("Desvio Padrão das Métricas GLCM – Comparação de Técnicas")
plt.xlabel("Métrica GLCM")
plt.ylabel("Desvio Padrão (%)")
plt.grid(True)
plt.legend()
plt.show()
```
